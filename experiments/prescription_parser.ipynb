### Image Processing

```python
import cv2
from PIL import Image
```

```python

img = cv2.imread('pre1.jpg', flags=cv2.IMREAD_GRAYSCALE)
img
```

```python
#Simple thresholding
_,new_image = cv2.threshold(img,150,255,cv2.THRESH_BINARY)


# 0 means black, 255 most white
# if value > 150 then value is 255, else = 0, 255 is max btw
```

```python
Image.fromarray(new_image).show()
```

```python
# Adaptive Thresholding
#adaptive thresholding gaussian -> divide bigger image into smaller regions and get 
# threshold to speciifc region

new_image = cv2.adaptiveThreshold(img,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                    cv2.THRESH_BINARY, 61,11)
Image.fromarray(new_image).show()
```

```python
import numpy as np
```

```python
def preprocess_image(img):
    gray = cv2.cvtColor(np.array(img),cv2.COLOR_BGR2GRAY)
    resized = cv2.resize(gray,None, fx=1.5,fy=1.5,interpolation=cv2.INTER_LINEAR)
    processed_image = cv2.adaptiveThreshold(resized,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                    cv2.THRESH_BINARY, 61,11)
    return processed_image
    
    
```

```python
img = preprocess_image(pages[0])
Image.fromarray(new_image).show()
```

```python
text = pytesseract.image_to_string(img, lang='eng')
print(text)
```

```python
import re
```

```python
text = 'Patient phone number is 7533876921.  (344)-111-3333 Bill amount www.12qqqw.com is 120$ 100 dollar'
```

```python
print(text)
```

```python
pattern_name = 'Name:(.*)Date:'
match = re.findall(pattern_name, text)
match[0].strip()
```

```python
pattern_address = 'Address:(.*)\n'
re.findall(pattern_address, text)
```

```python
pattern_med = "Address[^\n]*(.*)Directions"
match = re.findall(pattern_med, text,flags=re.DOTALL)
print(match[0].strip())
```

```python
pattern_dir = "Directions[^\n]*(.*)Refill"
match = re.findall(pattern_dir, text,flags=re.DOTALL)
print(match[0].strip())
```

```python
pattern_refill = "Refill:(.*)times"
match = re.findall(pattern_refill, text,flags=re.DOTALL)
print(match[0].strip())
```
